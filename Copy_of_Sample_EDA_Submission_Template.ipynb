### Dataset Loading

````python
# Load Dataset
!git clone https://github.com/PhonePe/pulse.git
````

````python
import os

path = "/content/pulse/data/aggregated/transaction/country/india/state/"
print(os.listdir(path))
````

````python
import json
import pandas as pd

data_list = []

for state in os.listdir(path):
    for year in os.listdir(path + state):
        for file in os.listdir(path + state + "/" + year):

            with open(path + state + "/" + year + "/" + file) as f:
                data = json.load(f)

                for item in data["data"]["transactionData"]:
                    data_list.append({
                        "state": state,
                        "year": int(year),
                        "quarter": int(file.strip(".json")),
                        "type": item["name"],
                        "count": item["paymentInstruments"][0]["count"],
                        "amount": item["paymentInstruments"][0]["amount"]
                    })

df = pd.DataFrame(data_list)
df.head()
````

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Make a copy of dataset
df_clean = df.copy()

# 1. Standardize state names
df_clean["state"] = df_clean["state"].str.replace("-", " ").str.title()

# 2. Check and remove duplicates
df_clean = df_clean.drop_duplicates()

# 3. Check missing values
print(df_clean.isnull().sum())

# 4. Convert data types (safety step)
df_clean["year"] = df_clean["year"].astype(int)
df_clean["quarter"] = df_clean["quarter"].astype(int)

# 5. Create new useful features

# Total transaction in lakhs (better readability)
df_clean["amount_lakhs"] = df_clean["amount"] / 100000

# Transactions per unit (efficiency metric)
df_clean["avg_transaction_value"] = df_clean["amount"] / df_clean["count"]

# 6. Sort values for analysis
df_clean = df_clean.sort_values(by=["year", "quarter"])

# Final check
df_clean.head()
````

Answer Here.
Several data preprocessing steps were performed to make the dataset analysis-ready.

1. State names were standardized by replacing hyphens with spaces and converting them into proper title case for better readability.

2. Duplicate records were checked and removed to ensure data integrity.

3. Missing values were analyzed, and it was found that the dataset contained no null values, indicating clean data.

4. Data types were verified and converted where necessary to maintain consistency.

5. New features were created to enhance analysis:
   - Transaction amount in lakhs for better visualization.
   - Average transaction value to understand user spending behavior.
